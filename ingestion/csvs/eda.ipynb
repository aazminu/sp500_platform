```python
import pandas as pd

def find_ticker_index_simple(ticker_symbol_to_find):
    """
    Loads 'tickers.csv' and finds the 0-based index of a ticker symbol
    in the unique list of tickers from the 'ticker' column.

    Args:
        ticker_symbol_to_find (str): The ticker symbol to search for.

    Returns:
        int: The 0-based index of the ticker.
    """
    # Load the CSV file
    tickers_df = pd.read_csv(
        "tickers.csv"
    )
    
    # Get unique tickers and convert to a list
    unique_tickers_list = list(tickers_df['ticker'].unique())
    
    # Find and return the index
    return unique_tickers_list.index(ticker_symbol_to_find)
ticker = 'TECH'
index = find_ticker_index_simple(ticker)
print(f"The ticker '{ticker}' is at index: {index}")
```
