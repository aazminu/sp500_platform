This file converts my csv training data file into a jsonl file for finetuning


```python
import pandas as pd
import json
```

```python
df = pd.read_csv("output.csv")
```

```python
with open("train.jsonl", "w", encoding="utf-8") as f:
    for _, row in df.iterrows():
        item = {
            "prompt": row["prompt"],
            "response": row["response"]
        }
        f.write(json.dumps(item) + "\n")
```

```python
import os
from dotenv import load_dotenv
from huggingface_hub import HfApi

# Load environment variables from .env file
load_dotenv()

# Get the token
hf_token = os.getenv("HF_TOKEN")

# Upload using HfApi
api = HfApi(token=hf_token)
api.upload_folder(
    folder_path="/path/to/local/dataset",  # <-- update this to your actual path
    repo_id="mdot77/sp500p",
    repo_type="dataset",
)
```
